```python


class ObjetoMochila:
    def __init__(self, nombre, valor, peso):
        self.nombre = nombre
        self.valor = valor
        self.peso = peso

        self.densidad = valor / peso

def mochila_fraccional(capacidad_maxima, nombres, valores, pesos):
    """
     problema de la mochila fraccional
    utilizando una estrategia Greedy basada en la densidad de valor.
    """
    total_elementos = len(nombres)
    lista_objetos = []


    for i in range(total_elementos):
        objeto_actual = ObjetoMochila(nombres[i], valores[i], pesos[i])
        lista_objetos.append(objeto_actual)


    def obtener_densidad(objeto):
        return objeto.densidad

    lista_objetos.sort(key=obtener_densidad, reverse=True)

    capacidad_actual = capacidad_maxima
    valor_total_obtenido = 0.0

    print("Proceso de selección (Mochila Fraccional)")
    for objeto in lista_objetos:
        if capacidad_actual <= 0:
            break


        if objeto.peso <= capacidad_actual:
            capacidad_actual = capacidad_actual - objeto.peso
            valor_total_obtenido = valor_total_obtenido + objeto.valor
            print(f" objeto {objeto.nombre} COMPLETO (Valor: {objeto.valor}, Peso: {objeto.peso})")
        else:

            fraccion_llevada = capacidad_actual / objeto.peso
            valor_parcial = objeto.valor * fraccion_llevada
            valor_total_obtenido = valor_total_obtenido + valor_parcial


            numerador, denominador = fraccion_llevada.as_integer_ratio()
            print(f"Se lleva {numerador}/{denominador} Parte del objeto {objeto.nombre} (Valor parcial: {valor_parcial:.1f})")

            capacidad_actual = 0

    print(f"Valor total aproximado obtenido: {valor_total_obtenido}\n")


def cobertura_antenas(casas, radio_cobertura):
    """
    problema de cobertura de antenas minimizando la cantidad
    de dispositivos mediante una estrategia Greedy de izquierda a derecha.
    """

    casas.sort()
    antenas_colocadas = []
    indice_casa = 0
    total_casas = len(casas)

    print(" Proceso de selección (Cobertura de Antenas)")
    while indice_casa < total_casas:

        casa_izquierda = casas[indice_casa]


        posicion_antena = casa_izquierda + radio_cobertura
        antenas_colocadas.append(posicion_antena)
        print(f"- Antena colocada en la posición: {posicion_antena}")


        limite_maximo_cobertura = posicion_antena + radio_cobertura

        while indice_casa < total_casas and casas[indice_casa] <= limite_maximo_cobertura:
            indice_casa = indice_casa + 1

    print(f"Cantidad total de antenas utilizadas: {len(antenas_colocadas)}\n")
    return antenas_colocadas



if __name__ == "__main__":
    print(" PRUEBA 1: MOCHILA FRACCIONAL (Ejemplo 1)")
    capacidad_mochila = 50
    nombres_objetos = ["A", "B", "C"]
    valores_objetos = [60, 100, 120]
    pesos_objetos = [10, 20, 30]

    mochila_fraccional(capacidad_mochila, nombres_objetos, valores_objetos, pesos_objetos)

    print("PRUEBA 2: COBERTURA DE ANTENAS (Ejemplo 1)")
    posiciones_casas = [1, 2, 7, 11, 20, 21, 30]
    radio_antena = 5

    cobertura_antenas(posiciones_casas, radio_antena)
```
